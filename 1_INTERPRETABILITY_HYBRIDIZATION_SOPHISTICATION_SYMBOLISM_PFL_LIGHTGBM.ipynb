```python
import numpy as np
import pandas as pd
import lightgbm as lgb
from sklearn.cluster import KMeans
from sklearn.metrics import (
    precision_recall_curve, auc, cohen_kappa_score, f1_score,
    recall_score, precision_score, accuracy_score, confusion_matrix,
    roc_auc_score, balanced_accuracy_score, matthews_corrcoef,
    log_loss, brier_score_loss, fbeta_score
)
from sklearn.calibration import CalibratedClassifierCV
from sklearn.isotonic import IsotonicRegression
import matplotlib.pyplot as plt
import seaborn as sns
import time
from pathlib import Path
import logging
import os
import sys
import traceback
import importlib.util
from typing import Dict, Tuple, List, Any, Optional
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold
from google.colab import drive
import warnings

# Configure the logging system
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    handlers=[
        logging.StreamHandler(),
        logging.FileHandler('/tmp/pfl_lightgbm.log')
    ]
)
logger = logging.getLogger(__name__)

# Definition of the PFLZ_CP class as a simple placeholder/interface.
# Its methods will be provided by the dynamically loaded PFL module.

class PFLZ_CP:
    def __init__(self, n_clusters: int, epsilon: float = 1e-8, lgb_params: Optional[Dict[str, Any]] = None, feature_names: Optional[List[str]] = None) -> None:
        self.n_clusters = n_clusters
        self.epsilon = epsilon
        self.lgb_params = lgb_params if lgb_params is not None else {}
        self.feature_names = feature_names
        self.cluster_centers_ = np.array([])
        self.cluster_model = None
        self.cond_probs_ = None
        self.n_original_features_ = 0
        self.n_fuzzy_features_ = n_clusters
        self.X_original_fit = None

    # All methods like calc_mf, fit, predict_proba, etc. ARE DEFINED IN THE PFL.
    # They are expected from the PFL module during dynamic loading.
    # If the PFL.py module is not loaded correctly, calls to these methods
    # on an instance of PFL_CP will raise an AttributeError, which is the desired behavior.

# Global variable to track if the actual PFL module has been successfully loaded
_PFLZ_MODULE_LOADED = False

# --- Module 1: Centralized Configuration ---

class SystemConfiguration:
    """Centralized configuration with validation and fixed clinical thresholds."""

    # Fixed system parameters
    NB_FEATURES = None
    NB_RULES = 7
    INDICES_FEATURES = None
    EPSILON = 1e-8

    # Fixed clinical thresholds
    CLINICAL_RISK_THRESHOLDS = {
        'LOW': 10.0,
        'MEDIUM': 15.0,
        'HIGH': 25.0
    }

    # LightGBM hyperparameters for consistency
    LGB_PARAMS = {
        'objective': 'binary',
        'metric': 'binary_logloss',
        'boosting_type': 'gbdt',
        'learning_rate': 0.08,
        'num_leaves': 25,
        'max_depth': 4,
        'min_data_in_leaf': 10,
        'feature_fraction': 0.8,
        'bagging_fraction': 0.85,
        'bagging_freq': 5,
        'lambda_l1': 0.3,
        'lambda_l2': 0.6,
        'min_gain_to_split': 0.02,
        'path_smooth': 100,
        'verbose': -1,
        'random_state': 42,
        'force_col_wise': True,
        'scale_pos_weight': 1
    }

    # Interpretability parameters
    COHERENCE_LAMBDA = 0.15
    COHERENCE_ALERT_THRESHOLD = 0.10

    PERCENTILE_COVERAGE = 60

    # Calibration parameters
    CALIBRATION_METHOD = 'isotonic'
    CALIBRATION_CV = 3

    # Store scalers for denormalization
    SCALERS = {}


    @classmethod
    def validate_configuration(cls):
        """Validates configuration consistency."""
        if cls.NB_FEATURES is not None and cls.INDICES_FEATURES is not None:
            if len(cls.INDICES_FEATURES) != cls.NB_FEATURES:
                raise ValueError("Inconsistency between NB_FEATURES and INDICES_FEATURES")
        logger.info("✅ Configuration validated")

# --- Module 2: Data Management ---

class DataManager:
    """Manages Drive mounting, locating, and loading data."""
    def __init__(self):
        self.mydrive_path = None
        self.modeling_data2_path = None
        self.all_feature_names = None
        # Attributes to store loaded DataFrames for easy access
        self.df_train = None
        self.X_train_df = None
        self.y_train_arr = None


    def mount_drive(self):
        """Mounts the Drive into the file system."""
        print("="*50)
        print("MOUNTING DRIVE")
        print("="*50)
        drive.mount('/content/drive', force_remount=True)
        self.mydrive_path = Path("/content/drive/MyDrive")
        print("Drive mounted successfully")

    def locate_data_folder(self):
        """Locates the 'modeling_data2' folder in the Drive."""
        print("\n" + "="*50)
        print("LOCATING DATA FOLDER")
        print("="*50)
        # Direct search for the 'modeling_data2' folder under 'PFL'
        pfl_dir_candidate = self.mydrive_path / "PFL"
        if pfl_dir_candidate.is_dir() and (pfl_dir_candidate / "modeling_data2").is_dir():
            self.modeling_data2_path = pfl_dir_candidate / "modeling_data2"
            print(f"✅ modeling_data2 folder found: {self.modeling_data2_path}")
            return

        # Fallback: general search if the direct path fails
        for root, dirs, _ in os.walk(self.mydrive_path):
            if "modeling_data2" in dirs:
                self.modeling_data2_path = Path(root) / "modeling_data2"
                print(f"✅ modeling_data2 folder found: {self.modeling_data2_path}")
                return
        raise FileNotFoundError("modeling_data2 folder not found in Drive")

    def load_data(self):
        """Loads training data (df_train, X_train, y_train) from the folder."""
        print("\n" + "="*50)
        print("LOADING DATA")
        print("="*50)
        df_train_path = self.modeling_data2_path / 'df_train.hdf'
        if not df_train_path.exists():
            raise FileNotFoundError(f"File not found: {df_train_path}")
        self.df_train = pd.read_hdf(df_train_path, key='Processed_data')

        X_train_df_temp = pd.read_hdf(self.modeling_data2_path / 'X_train.hdf', key='Processed_data')
        self.all_feature_names = [col for col in X_train_df_temp.columns if col != 'A']

        A_train = self.df_train['A'].values
        self.X_train_df = pd.read_hdf(self.modeling_data2_path / 'X_train.hdf', key='Processed_data')
        self.X_train_df['A'] = A_train

        self.y_train_arr = pd.read_hdf(self.modeling_data2_path / 'y_train.hdf', key='Processed_data').values.ravel()

        # --- Calculation and storage of true min/max values ---
        print("\n" + "="*50)
        print("CALCULATING AND STORING REAL MIN/MAX VALUES FOR NORMALIZATION")
        print("="*50)

        real_min_max_values = {}
        # Iterate over self.all_feature_names which already contains all feature names except 'A'
        for feature_name in self.all_feature_names:
            if feature_name in self.X_train_df.columns and pd.api.types.is_numeric_dtype(self.X_train_df[feature_name]):
                min_val = self.X_train_df[feature_name].min()
                max_val = self.X_train_df[feature_name].max()
                real_min_max_values[feature_name] = (min_val, max_val)
                print(f"Feature '{feature_name}': Min = {min_val:.4f}, Max = {max_val:.4f}")
            else:
                print(f"Warning: Feature '{feature_name}' is not numeric or was not found in X_train_df for min/max calculation.")


        SystemConfiguration.SCALERS['min_max'] = real_min_max_values

        print("\n🌟 Data loaded successfully!")
        return self.X_train_df, self.y_train_arr, self.all_feature_names

# --- Module 3: PFL Module Loader ---

class PFLModuleLoader:
    """Manages the dynamic loading of the PFLZ class (alias PFLZ_CP) from its external file
    ensuring it is correctly imported."""
    @staticmethod
    def load_pflz_cp_module(mydrive_path: Path):
        print("\n" + "="*50)
        print("LOADING PFL MODULE")
        print("="*50)
        pflz_file_path = None
        # Searches for the PFLZ.py file in the Modules directory
        candidate_path = mydrive_path / "PFL" / "Modules" / "PFLZ.py"
        if candidate_path.exists():
            pflz_file_path = candidate_path
        else:
            # Fallback: broader search if the direct path fails
            for root, dirs, files in os.walk(mydrive_path):
                if "PFL" in dirs:
                    pfl_dir = Path(root) / "PFL"
                    if "Modules" in os.listdir(pfl_dir):
                        modules_path = pfl_dir / "Modules"
                        if "PFLZ.py" in os.listdir(modules_path):
                            pflz_file_path = modules_path / "PFLZ.py"
                            break

        if pflz_file_path is None:
            raise FileNotFoundError("PFLZ.py file not found in Drive. Please ensure PFLZ.py exists in the 'PFL/Modules' folder.")

        print(f"✅ PFLZ.py file found: {pflz_file_path}")

        # We directly read the file as a Python module
        spec = importlib.util.spec_from_file_location("PFLZ_module_temp", pflz_file_path)
        if spec is None:
            raise ImportError(f"Could not specify module for {pflz_file_path}.")

        module = importlib.util.module_from_spec(spec)
        sys.modules["PFLZ_module_temp"] = module
        spec.loader.exec_module(module)

        loaded_pflz_class = getattr(module, 'PFLZ', None)

        if loaded_pflz_class is None:
            raise ImportError("The class 'PFLZ' could not be found in the loaded module. Check the class name in PFLZ.py.")

        class PFLZ_CP_Alias(loaded_pflz_class):
            def __init__(self, *args, **kwargs):
                super().__init__(*args, **kwargs)
                if not hasattr(self, 'X_original_fit'):
                    self.X_original_fit = None

        return PFLZ_CP_Alias

# --- Module 4: Interpretability Calculations ---

class InterpretabilityCalculator:
    """Calculates interpretability scores and generates clinical rules from the PFLZ model."""

    @staticmethod
    def calculate_interpretability_features(pflz_cp, X: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, Dict, Dict]:
        """
        Calculates interpretability features (scores, confidence, dominant clusters)
        and cluster interpretations (fuzzy rules).
        """
        logger.info("Calculating interpretability features...")

        try:
            membership_values = pflz_cp.calc_mf(X, normalized=True)
            if membership_values is None or membership_values.size == 0 or membership_values.shape[1] == 0:
                raise ValueError("Invalid or empty fuzzy membership values after calc_mf.")
        except Exception as e:
            logger.error(f"Error calculating membership values: {e}")
            raise

        try:
            conditional_probs = pflz_cp.calc_cond_probs(X, y)
            if conditional_probs is None or conditional_probs.size == 0 or conditional_probs.shape[1] < 2:
                raise ValueError("Invalid or empty conditional probabilities after calc_cond_probs.")
        except Exception as e:
            logger.error(f"Error calculating conditional probabilities: {e}")
            raise

        n_samples = X.shape[0]
        raw_interpretability_scores = np.zeros(n_samples)
        dominant_clusters = np.zeros(n_samples, dtype=int)
        confidence_scores = np.zeros(n_samples)

        num_clusters_actual = min(pflz_cp.n_clusters, membership_values.shape[1])

        for i in range(n_samples):
            membership_i = membership_values[i, :num_clusters_actual]
            weighted_probs = membership_i * conditional_probs[:num_clusters_actual, 1]

            if weighted_probs.size > 0:
                max_weighted_prob = np.max(weighted_probs)
                raw_interpretability_scores[i] = max_weighted_prob
                dominant_clusters[i] = np.argmax(weighted_probs)
            else:
                raw_interpretability_scores[i] = 0.0
                dominant_clusters[i] = -1

            confidence_scores[i] = np.max(membership_i) if membership_i.size > 0 else 0.0

        if np.max(raw_interpretability_scores) > np.min(raw_interpretability_scores):
            normalized_interpretability_scores = (
                (raw_interpretability_scores - np.min(raw_interpretability_scores)) /
                (np.max(raw_interpretability_scores) - np.min(raw_interpretability_scores))
            )
        else:
            normalized_interpretability_scores = raw_interpretability_scores

        adaptive_threshold = np.percentile(normalized_interpretability_scores,
                                             SystemConfiguration.PERCENTILE_COVERAGE)

        diagnostics = InterpretabilityCalculator._diagnose_distributions(
            raw_interpretability_scores,
            normalized_interpretability_scores,
            adaptive_threshold
        )

        interpretability_features = np.column_stack([
            normalized_interpretability_scores,
            confidence_scores,
            dominant_clusters
        ])

        cluster_interpretations = {}
        if hasattr(pflz_cp, 'cluster_centers_') and pflz_cp.cluster_centers_.shape[0] > 0:
            rule_widths = pflz_cp.calc_rule_widths()
        else:
            logger.warning("PFLZ_CP centroids are not available for rule width calculation. Using default widths.")
            num_features = X.shape[1] if X.shape[0] > 0 else 1
            rule_widths = np.full((pflz_cp.n_clusters, num_features), 0.5)

        alpha_rule = 1.5
        for cluster_id in range(num_clusters_actual):
            if hasattr(pflz_cp, 'cluster_centers_') and pflz_cp.cluster_centers_.shape[0] > cluster_id and pflz_cp.feature_names is not None:
                cluster_center = pflz_cp.cluster_centers_[cluster_id]
                denormalized_rule = InterpretabilityCalculator._generate_centroid_rule(
                    cluster_center,
                    pflz_cp.feature_names,
                    SystemConfiguration.SCALERS.get('min_max', {}),
                    alpha=alpha_rule,
                    widths=rule_widths[cluster_id] if rule_widths.shape[0] > cluster_id else np.full(cluster_center.shape[0], 0.5)
                )
                cluster_interpretations[cluster_id] = {
                    'conditional_prob': conditional_probs[cluster_id, 1],
                    'rule': denormalized_rule
                }
            else:
                logger.warning(f"Cluster {cluster_id} or feature names are not available. Skipping rule generation for this cluster.")

        logger.info(f"✅ Interpretability features calculated - Adaptive threshold: {adaptive_threshold:.3f}")

        return interpretability_features, cluster_interpretations, diagnostics

    @staticmethod
    def _diagnose_distributions(raw_scores: np.ndarray, normalized_scores: np.ndarray,
                                   adaptive_threshold: float) -> Dict:
        """Provides a detailed diagnosis of interpretability score distributions."""
        diagnostics = {
            'raw_scores_stats': {
                'min': np.min(raw_scores), 'max': np.max(raw_scores),
                'mean': np.mean(raw_scores), 'std': np.std(raw_scores),
                'percentiles': np.percentile(raw_scores, [25, 50, 75, 90, 95])
            },
            'normalized_scores_stats': {
                'min': np.min(normalized_scores), 'max': np.max(normalized_scores),
                'mean': np.mean(normalized_scores), 'std': np.std(normalized_scores),
                'percentiles': np.percentile(normalized_scores, [25, 50, 75, 90, 95])
            },
            'adaptive_threshold': adaptive_threshold,
            'adaptive_coverage': np.mean(normalized_scores > adaptive_threshold)
        }

        diagnostics['normalized_interpretability_scores_original'] = normalized_scores
        logger.info(f"Diagnosis - Adaptive coverage: {diagnostics['adaptive_coverage']:.1%}")
        return diagnostics

    @staticmethod
    def _generate_centroid_rule(centroid: np.ndarray, feature_names: List[str], scalers: Dict = None, alpha: float = 1.0, widths: Optional[np.ndarray] = None) -> str:
        """
        Generates a readable clinical rule from a centroid, with denormalization.
        Uses `feature_names` directly.
        Alpha is used to adjust the interval width.
        `widths` provides specific widths for each centroid feature.
        """
        if scalers is None:
            scalers = {}

        conditions = []
        num_features_to_process = min(len(centroid), len(feature_names))
        if len(centroid) != len(feature_names):
            logger.warning(f"Inconsistency: Centroid has {len(centroid)} elements but {len(feature_names)} feature names. Processing the first {num_features_to_process}.")

        if widths is None or len(widths) != num_features_to_process:
            logger.warning(f"Widths not provided or incorrect dimension for _generate_centroid_rule. Using a default width.")
            default_width = 0.5
            widths = np.full(num_features_to_process, default_width)

        for i in range(num_features_to_process):
            original_feature_name = feature_names[i]
            normalized_value = centroid[i]
            feature_width = widths[i]

            lower_bound_norm = normalized_value - alpha * feature_width
            upper_bound_norm = normalized_value + alpha * feature_width

            if 'min_max' in scalers and original_feature_name in scalers['min_max']:
                original_min, original_max = scalers['min_max'][original_feature_name]
                lower_bound_denorm = lower_bound_norm * (original_max - original_min) + original_min
                upper_bound_denorm = upper_bound_norm * (original_max - original_min) + original_min
                conditions.append(f"{original_feature_name} ∈ [{lower_bound_denorm:.2f}, {upper_bound_denorm:.2f}]")
            else:
                conditions.append(f"{original_feature_name} ∈ [{lower_bound_norm:.3f}, {upper_bound_norm:.3f}]")

        simplified_rule = " AND ".join(conditions)

        return simplified_rule

# --- Module 5: Metrics with Calibration ---

class MetricsCalculator:
    """Calculates performance metrics and manages post-training calibration."""

    @staticmethod
    def calculate_ece(y_true: np.ndarray, y_pred_proba: np.ndarray, n_bins: int = 10) -> float:
        """Calculates Expected Calibration Error (ECE)."""
        bin_edges = np.linspace(0, 1, n_bins + 1)
        bin_indices = np.digitize(y_pred_proba, bin_edges[:-1])
        ece = 0.0
        for i in range(1, n_bins + 1):
            in_bin = bin_indices == i
            bin_count = np.sum(in_bin)
            if bin_count > 0:
                bin_true_mean = np.mean(y_true[in_bin])
                bin_pred_mean = np.mean(y_pred_proba[in_bin])
                ece += np.abs(bin_true_mean - bin_pred_mean) * (bin_count / len(y_true))
        return ece

    @staticmethod
    def apply_post_training_calibration(y_true: np.ndarray, y_pred_proba: np.ndarray,
                                         method: str = 'isotonic') -> Tuple[np.ndarray, object]:
        """Applies post-training calibration to improve the reliability of probabilities."""
        logger.info(f"Applying {method} calibration...")
        if method == 'isotonic':
            calibrator = IsotonicRegression(out_of_bounds='clip')
        elif method == 'sigmoid':
            calibrator = LogisticRegression(solver='lbfgs')
        else:
            raise ValueError(f"Unknown calibration method: {method}")

        if method == 'sigmoid':
            logits = np.log(np.clip(y_pred_proba, 1e-15, 1-1e-15) /
                            np.clip(1-y_pred_proba, 1e-15, 1-1e-15))
            calibrator.fit(logits.reshape(-1, 1), y_true)
            y_pred_calibrated = calibrator.predict_proba(logits.reshape(-1, 1))[:, 1]
        else:
            calibrator.fit(y_pred_proba, y_true)
            y_pred_calibrated = calibrator.predict(y_pred_proba)

        logger.info("✅ Calibration applied")
        return y_pred_calibrated, calibrator

# --- Module 6: Training with Calibration and Constraints ---

class ModelTrainer:
    """Trains the tree-based model with calibration and integrates interpretability features."""

    def __init__(self, config: SystemConfiguration):
        self.config = config
        self.config.validate_configuration()
        self.calibrators = {}

    def train_and_evaluate(self, X_subset: pd.DataFrame, y_subset: np.ndarray,
                                   subset_name: str, selected_features: List[str]) -> Dict:
        """Trains and evaluates the tree-based model for a given subset."""
        logger.info(f"Starting processing for {subset_name} with the tree-based model.")
        start_time = time.time()

        global _PFLZ_MODULE_LOADED
        if not _PFLZ_MODULE_LOADED:
            raise RuntimeError("PFL module was not loaded correctly. Cannot continue training.")

        try:
            pflz_cp = PFLZ_CP(
                n_clusters=self.config.NB_RULES,
                epsilon=self.config.EPSILON,
                lgb_params=self.config.LGB_PARAMS,
                feature_names=selected_features
            )
            pflz_cp.fit(X_subset.values, y_subset)
            pflz_cp.X_original_fit = X_subset.values

        except Exception as e:
            logger.error(f"Error during PFLZ_CP initialization or fitting for {subset_name}: {e}")
            raise

        centroids = pflz_cp.cluster_centers_

        interp_features, cluster_interpretations, diagnostics = (
            InterpretabilityCalculator.calculate_interpretability_features(
                pflz_cp, X_subset.values, y_subset
            )
        )

        self._display_diagnostics(diagnostics, subset_name)
        self._display_cluster_interpretations(cluster_interpretations)

        results_train_eval = self._cross_validation(
            X_subset.values, y_subset,
            pflz_cp, interp_features, subset_name, diagnostics
        )

        self._display_results(results_train_eval, subset_name)

        results_train_eval['time_minutes'] = round((time.time() - start_time) / 60, 2)
        logger.info(f"✅ Processing for {subset_name} completed in {results_train_eval['time_minutes']} min for the tree-based model.")

        return {
            'centroids': centroids,
            'predictions': results_train_eval['y_pred_proba_calibrated'],
            'raw_predictions': results_train_eval['y_pred_proba'],
            'results': results_train_eval,
            'pflz_cp': pflz_cp,
            'interpretations': cluster_interpretations,
            'diagnostics': diagnostics,
            'calibrator': self.calibrators.get(subset_name)
        }

    def _cross_validation(self, X: np.ndarray, y: np.ndarray, pflz_cp: Any,
                                   interp_features: np.ndarray, subset_name: str,
                                   diagnostics: Dict) -> Dict:
        """Performs cross-validation of the tree-based model with calibration and feature weighting."""
        logger.info(f"Starting cross-validation for the tree-based model ({subset_name}).")

        skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

        metrics = {
            'accuracy': [], 'balanced_accuracy': [], 'kappa': [], 'mcc': [],
            'precision': [], 'recall': [], 'f1_score': [], 'specificity': [], 'pr_auc': [], 'roc_auc': [], 'log_loss': [], 'brier': [],
            'fdr': [], 'for': []
        }

        y_pred_proba_full = np.zeros(len(y))
        y_pred_full = np.zeros(len(y))
        coherence_gaps_fold = []

        unique_classes, counts = np.unique(y, return_counts=True)
        if len(unique_classes) > 1:
            majority_class_count = counts[0] if unique_classes[0] == 0 else counts[1]
            minority_class_count = counts[1] if unique_classes[1] == 1 else counts[0]
            scale_pos_weight_val = majority_class_count / minority_class_count if minority_class_count > 0 else 1
        else:
            scale_pos_weight_val = 1

        for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
            logger.debug(f"Processing fold {fold+1}/5 for the tree-based model.")

            X_train_fold = X[train_idx]
            X_val_fold = X[val_idx]
            y_train_fold = y[train_idx]
            y_val_fold = y[val_idx]

            fuzzy_train = pflz_cp.calc_mf(X_train_fold, normalized=True)
            fuzzy_val = pflz_cp.calc_mf(X_val_fold, normalized=True)

            interp_train = interp_features[train_idx]
            interp_val = interp_features[val_idx]

            interpretability_weight = 15.0

            weighted_interp_train = interp_train * interpretability_weight
            weighted_interp_val = interp_val * interpretability_weight

            if fuzzy_train.shape[1] == 0 and self.config.NB_RULES > 0:
                fuzzy_train = np.zeros((X_train_fold.shape[0], self.config.NB_RULES))
            if fuzzy_val.shape[1] == 0 and self.config.NB_RULES > 0:
                fuzzy_val = np.zeros((X_val_fold.shape[0], self.config.NB_RULES))

            if weighted_interp_train.shape[1] == 0:
                weighted_interp_train = np.zeros((X_train_fold.shape[0], 3))
            if weighted_interp_val.shape[1] == 0:
                weighted_interp_val = np.zeros((X_val_fold.shape[0], 3))

            X_train_extended = np.hstack((X_train_fold, fuzzy_train, weighted_interp_train))
            X_val_extended = np.hstack((X_val_fold, fuzzy_val, weighted_interp_val))

            lgb_params_fold = self.config.LGB_PARAMS.copy()
            lgb_params_fold['scale_pos_weight'] = scale_pos_weight_val

            train_data = lgb.Dataset(X_train_extended, label=y_train_fold, free_raw_data=False)
            val_data = lgb.Dataset(X_val_extended, label=y_val_fold, reference=train_data, free_raw_data=False)

            model = lgb.train(
                lgb_params_fold,
                train_data,
                num_boost_round=500,
                valid_sets=[val_data],
                callbacks=[
                    lgb.early_stopping(stopping_rounds=30, verbose=False),
                ]
            )

            y_pred_proba = model.predict(X_val_extended)
            y_pred = (y_pred_proba > 0.2).astype(int)

            y_pred_proba_full[val_idx] = y_pred_proba
            y_pred_full[val_idx] = y_pred

            if interp_val.shape[1] > 0:
                fold_gap = np.mean(np.abs(y_pred_proba - interp_val[:, 0]))
            else:
                fold_gap = 1.0
            coherence_gaps_fold.append(fold_gap)

            self._calculate_fold_metrics(y_val_fold, y_pred, y_pred_proba, metrics)

        y_pred_proba_calibrated, calibrator = MetricsCalculator.apply_post_training_calibration(
            y, y_pred_proba_full, method=self.config.CALIBRATION_METHOD
        )
        self.calibrators[subset_name] = calibrator

        ece_raw = MetricsCalculator.calculate_ece(y, y_pred_proba_full)
        ece_calibrated = MetricsCalculator.calculate_ece(y, y_pred_proba_calibrated)

        return {
            'metrics': metrics,
            'y_pred_proba': y_pred_proba_full,
            'y_pred_proba_calibrated': y_pred_proba_calibrated,
            'y_pred': y_pred_full,
            'ece_raw': ece_raw,
            'ece_calibrated': ece_calibrated,
            'ece_improvement': ece_raw - ece_calibrated,
            'coherence_gaps_folds': coherence_gaps_fold,
            'mean_coherence_gap': np.mean(coherence_gaps_fold)
        }

    def _calculate_fold_metrics(self, y_true: np.ndarray, y_pred: np.ndarray, y_pred_proba: np.ndarray, metrics: Dict):
        """Calculates performance metrics for a single fold, handling single-class cases."""
        try:
            y_pred_proba = np.clip(y_pred_proba, 1e-15, 1 - 1e-15)

            if len(np.unique(y_true)) == 1 or len(np.unique(y_pred)) == 1:
                tn, fp, fn, tp = (0,0,0,0)
                if len(np.unique(y_true)) == 1 and y_true[0] == 0:
                    tn = np.sum(y_pred == 0)
                    fp = np.sum(y_pred == 1)
                elif len(np.unique(y_true)) == 1 and y_true[0] == 1:
                    tp = np.sum(y_pred == 1)
                    fn = np.sum(y_pred == 0)
                else:
                    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
                    tn, fp, fn, tp = cm.ravel()
            else:
                cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
                tn, fp, fn, tp = cm.ravel()

            specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
            fdr = fp / (tp + fp) if (tp + fp) > 0 else 0
            for_rate = fn / (tn + fn) if (tn + fn) > 0 else 0

            metrics['accuracy'].append(accuracy_score(y_true, y_pred))
            metrics['balanced_accuracy'].append(balanced_accuracy_score(y_true, y_pred))
            metrics['kappa'].append(cohen_kappa_score(y_true, y_pred))
            metrics['mcc'].append(matthews_corrcoef(y_true, y_pred))
            metrics['precision'].append(precision_score(y_true, y_pred, zero_division=0))
            metrics['recall'].append(recall_score(y_true, y_pred, zero_division=0))
            metrics['f1_score'].append(f1_score(y_true, y_pred, zero_division=0))
            metrics['specificity'].append(specificity)

            if len(np.unique(y_true)) > 1:
                precision_vals, recall_vals, _ = precision_recall_curve(y_true, y_pred_proba)
                metrics['pr_auc'].append(auc(recall_vals, precision_vals))
                metrics['roc_auc'].append(roc_auc_score(y_true, y_pred_proba))
            else:
                metrics['pr_auc'].append(np.nan)
                metrics['roc_auc'].append(np.nan)

            metrics['log_loss'].append(log_loss(y_true, y_pred_proba))
            metrics['brier'].append(brier_score_loss(y_true, y_pred_proba))
            metrics['fdr'].append(fdr)
            metrics['for'].append(for_rate)

        except Exception as e:
            logger.error(f"Error calculating metrics for a fold: {e}. Filling with 0.0 or NaN.")
            for key in metrics.keys():
                if key in ['pr_auc', 'roc_auc']:
                    metrics[key].append(np.nan)
                else:
                    metrics[key].append(0.0)

    def _display_diagnostics(self, diagnostics: Dict, subset_name: str):
        """Displays detailed interpretability diagnostics."""
        print(f"\n🌟 INTERPRETABILITY DIAGNOSTICS FOR {subset_name}")
        print("-" * 60)
        print(f"Raw Scores - Min: {diagnostics['raw_scores_stats']['min']:.4f}, "
              f"Max: {diagnostics['raw_scores_stats']['max']:.4f}, "
              f"Mean: {diagnostics['raw_scores_stats']['mean']:.4f}")
        print(f"Normalized Scores - Min: {diagnostics['normalized_scores_stats']['min']:.4f}, "
              f"Max: {diagnostics['normalized_scores_stats']['max']:.4f}, "
              f"Mean: {diagnostics['normalized_scores_stats']['mean']:.4f}")
        print(f"Adaptive Threshold ({SystemConfiguration.PERCENTILE_COVERAGE}th percentile): {diagnostics['adaptive_threshold']:.4f}")
        print(f"Adaptive Coverage: {diagnostics['adaptive_coverage']:.1%}")

    def _display_cluster_interpretations(self, cluster_interpretations: Dict):
        """Displays cluster interpretations (associated fuzzy rules)."""
        print("\n🌟 Cluster-based Interpretations:")
        for cluster_id, interpretation in cluster_interpretations.items():
            print(f"\nCluster {cluster_id}:")
            print(f"    • P(mortality|cluster) = {interpretation['conditional_prob']:.2%}")
            print(f"    • Rule: {interpretation['rule']}")

    def _display_results(self, results: Dict, subset_name: str):
        """Displays performance and calibration metrics of the tree-based model."""
        print(f"\n{'-'*70}")
        print(f"METRICS FOR {subset_name} (Tree-based model)")
        print(f"{'-'*70}")

        mean_metrics = {
            'Accuracy': np.nanmean(results['metrics']['accuracy']) * 100,
            'Balanced Accuracy': np.nanmean(results['metrics']['balanced_accuracy']) * 100,
            'Kappa': np.nanmean(results['metrics']['kappa']),
            'MCC': np.nanmean(results['metrics']['mcc']),
            'Precision': np.nanmean(results['metrics']['precision']) * 100,
            'Recall': np.nanmean(results['metrics']['recall']) * 100,
            'F1-Score': np.nanmean(results['metrics']['f1_score']) * 100,
            'Specificity': np.nanmean(results['metrics']['specificity']) * 100,
            'PR-AUC': np.nanmean(results['metrics']['pr_auc']),
            'ROC-AUC': np.nanmean(results['metrics']['roc_auc']),
            'Log Loss': np.nanmean(results['metrics']['log_loss']),
            'Brier': np.nanmean(results['metrics'].get('brier', [np.nan])),
            'FDR': np.nanmean(results['metrics']['fdr']),
            'FOR': np.nanmean(results['metrics']['for'])
        }

        calibration_metrics = {
            'Raw ECE': results.get('ece_raw', np.nan),
            'Calibrated ECE': results.get('ece_calibrated', np.nan),
            'ECE Improvement': results.get('ece_improvement', np.nan),
        }

        print("🌟 PERFORMANCE METRICS of the tree-based model:")
        print("Accuracy                                  : {:.2f}%".format(mean_metrics['Accuracy']))
        print("Balanced Accuracy                         : {:.2f}%".format(mean_metrics['Balanced Accuracy']))
        print("Kappa                                     : {:.4f}".format(mean_metrics['Kappa']))
        print("MCC                                       : {:.4f}".format(mean_metrics['MCC']))
        print("Precision                                 : {:.2f}%".format(mean_metrics['Precision']))
        print("Recall                                    : {:.2f}%".format(mean_metrics['Recall']))
        print("F1-Score                                  : {:.2f}%".format(mean_metrics['F1-Score']))
        print("Specificity                               : {:.2f}%".format(mean_metrics['Specificity']))
        print("PR-AUC                                    : {:.4f}".format(mean_metrics['PR-AUC'] if not np.isnan(mean_metrics['PR-AUC']) else 0.0))
        print("ROC-AUC                                   : {:.4f}".format(mean_metrics['ROC-AUC'] if not np.isnan(mean_metrics['ROC-AUC']) else 0.0))
        print("Log Loss                                  : {:.4f}".format(mean_metrics['Log Loss']))
        print("Brier Score                               : {:.4f}".format(mean_metrics['Brier'] if not np.isnan(mean_metrics['Brier']) else 0.0))
        print("FDR                                       : {:.4f}".format(mean_metrics['FDR']))
        print("FOR                                       : {:.4f}".format(mean_metrics['FOR']))

        print("\n🌟 CALIBRATION METRICS of the tree-based model:")
        for name, value in calibration_metrics.items():
            if 'ECE' in name:
                print(f"    {name:<20}: {value:.4f}")
            else:
                print(f"    {name:<20}: {value:.4f}")

        print(f"{'-'*70}")

# --- Module 7: Risk Cluster Analyzer ---

class RiskClusterAnalyzer:
    """Analyzes risk clusters with fixed clinical thresholds and coherence alerts."""

    @staticmethod
    def analyze_risk_clusters(X: pd.DataFrame, centroids: np.ndarray,
                                     predictions: np.ndarray, subset_name: str,
                                     selected_features: List[str],
                                     pflz_cp: Any, cluster_interpretations: Dict,
                                     raw_predictions: Optional[np.ndarray] = None,
                                     y_true: Optional[np.ndarray] = None,
                                     normalized_interp_scores: Optional[np.ndarray] = None):
        """
        Performs a detailed analysis by cluster, classifying risk based on fixed clinical thresholds
        and evaluating coherence between the tree-based model and fuzzy rules.
        """
        logger.info(f"Analyzing risk clusters for {subset_name}")

        if centroids.shape[0] == 0 or X.empty or len(selected_features) == 0:
            logger.warning(f"Centroids, X data, or selected features are empty for risk cluster analysis of {subset_name}. Skipping detailed cluster analysis.")
            return {'coherence_alerts': []}

        valid_features_for_clustering = [f for f in selected_features if f in X.columns]
        if not valid_features_for_clustering:
            logger.warning(f"None of the selected features ({selected_features}) are present in the X DataFrame. Cannot perform clustering for {subset_name}. Skipping.")
            return {'coherence_alerts': []}

        km = KMeans(n_clusters=len(centroids), init=centroids, n_init=1, random_state=42)
        clusters = km.fit_predict(X[valid_features_for_clustering].values)

        print(f"\n{'-'*70}")
        print(f"RISK CLUSTERS FOR {subset_name}")
        print(f"{'-'*70}")

        thresholds = SystemConfiguration.CLINICAL_RISK_THRESHOLDS
        print("🌟 FIXED CLINICAL THRESHOLDS:")
        print(f"- LOW RISK: < {thresholds['LOW']:.1f}%")
        print(f"- MEDIUM RISK: {thresholds['LOW']:.1f}% to {thresholds['MEDIUM']:.1f}%")
        print(f"- HIGH RISK: ≥ {thresholds['HIGH']:.1f}%")
        print(f"{'-'*70}")

        global_mean_risk = np.mean(predictions) * 100
        print(f"🌟 Global mean risk of the subset: {global_mean_risk:.1f}%")

        cluster_coherence_alerts = []

        for i in range(len(centroids)):
            cluster_mask = (clusters == i)
            n_patients = np.sum(cluster_mask)

            if n_patients == 0:
                print(f"\nCluster {i}: No patients")
                continue

            if i not in cluster_interpretations:
                logger.warning(f"Cluster ID {i} was not found in 'cluster_interpretations'. Skipping this cluster for detailed analysis.")
                continue

            tree_model_mean_risk = np.mean(predictions[cluster_mask]) * 100

            if raw_predictions is not None:
                tree_model_raw_risk = np.mean(raw_predictions[cluster_mask]) * 100
                calibration_improvement = abs(tree_model_raw_risk - tree_model_mean_risk)
            else:
                tree_model_raw_risk = tree_model_mean_risk
                calibration_improvement = 0

            if tree_model_mean_risk < thresholds['LOW']:
                risk_level = "LOW"
                probable_outcome = "Probable Survival"
                emoji_color = "🌟"
            elif tree_model_mean_risk <= thresholds['MEDIUM']:
                risk_level = "MEDIUM"
                probable_outcome = "Enhanced Surveillance"
                emoji_color = "✨"
            else:
                risk_level = "HIGH"
                probable_outcome = "Urgent Intervention"
                emoji_color = "🚨"

            conditional_prob = cluster_interpretations[i].get('conditional_prob', 0.0)
            simplified_rule = cluster_interpretations[i].get('rule', 'Rule not available')

            cluster_coherence_gap = abs(tree_model_mean_risk/100 - conditional_prob)

            specific_coherence_alert = cluster_coherence_gap > SystemConfiguration.COHERENCE_ALERT_THRESHOLD
            if specific_coherence_alert:
                cluster_coherence_alerts.append({
                    'cluster': i,
                    'gap': cluster_coherence_gap,
                    'tree_model_risk': tree_model_mean_risk,
                    'rules_prob': conditional_prob * 100
                })

            cluster_fidelity = 1 - cluster_coherence_gap
            cluster_fidelity = max(0, min(1, cluster_fidelity))

            print(f"\n{emoji_color} Cluster {i} - {risk_level.upper()} RISK")
            print(f"    • Calibrated risk (Tree-based model): {tree_model_mean_risk:.1f}%")
            if raw_predictions is not None:
                print(f"    • Raw risk (Tree-based model): {tree_model_raw_risk:.1f}% (improvement: {calibration_improvement:.1f}%)")
            print(f"    • P(mortality|cluster) (Fuzzy Rule): {conditional_prob:.2%}")
            print(f"    • Cluster Fidelity                                   : {cluster_fidelity:.4f}")
            print(f"    • Clinical Recommendation: {probable_outcome}")
            print(f"    • Number of patients: {n_patients:,}")
            print(f"    • Percentage of subset: {(n_patients/len(predictions))*100:.1f}%")

            print(f"    • Clinical rule: {simplified_rule}")

            coherence_status_display = ""
            if cluster_coherence_gap < (SystemConfiguration.COHERENCE_ALERT_THRESHOLD / 2):
                coherence_status_display = "✅ Excellent"
            elif cluster_coherence_gap < SystemConfiguration.COHERENCE_ALERT_THRESHOLD:
                coherence_status_display = "✨ Good"
            else:
                coherence_status_display = "🚨 Poor - In-depth analysis required"

            print(f"    • Rule-tree model coherence: {coherence_status_display} (gap: {cluster_coherence_gap:.1%})")

            if specific_coherence_alert:
                print(f"    🚨 ALERT: Significant gap detected - For this cluster, consider:")
                print("      - A thorough review of feature interactions.")
                print("      - Validation with clinical expertise for these patient profiles.")

            print(f"{'-'*70}")

        if cluster_coherence_alerts:
            print(f"\n🚨 SUMMARY OF CLUSTER COHERENCE ALERTS ({len(cluster_coherence_alerts)} cluster(s)):")
            for alert in cluster_coherence_alerts:
                print(f"    • Cluster {alert['cluster']}: Tree-based model={alert['tree_model_risk']:.1f}% vs Rules={alert['rules_prob']:.1f}% (Gap: {alert['gap']:.1%})")
            print("\n✨ General Recommendations for Discrepancies:")
            print("    - **Rule Review:** Collaborate with clinical expertise to refine or create more precise rules for these patient profiles.")
            print("    - **Feature Enrichment:** Consider adding additional features to the tree-based model.")
        else:
            print(f"\n✅ SATISFACTORY CLUSTER COHERENCE: No significant gaps detected.")

        print("\n📈 VISUALIZATIONS OF RESULTS:")
        plt.figure(figsize=(15, 6))

        if y_true is not None and normalized_interp_scores is not None and len(predictions) > 0:
            plt.subplot(1, 2, 1)
            sns.histplot(predictions, color='skyblue', label='Tree-based Model Predictions (Calibrated)', kde=True, stat='density')
            sns.histplot(normalized_interp_scores, color='salmon', label='Interpretability Scores (Rules)', kde=True, stat='density')
            plt.title(f'Distribution of Scores and Predictions for {subset_name}')
            plt.xlabel('Probability / Score')
            plt.ylabel('Density')
            plt.legend()
            plt.grid(True, linestyle='--', alpha=0.6)

            plt.subplot(1, 2, 2)
            bin_middles = np.linspace(0.05, 0.95, 10)
            fraction_of_positives = []
            mean_predicted_value = []

            if len(y_true) == len(predictions):
                for i in range(10):
                    low = i / 10.0
                    high = (i + 1) / 10.0
                    mask = (predictions >= low) & (predictions < high)
                    if np.any(mask):
                        fraction_of_positives.append(np.mean(y_true[mask]))
                        mean_predicted_value.append(np.mean(predictions[mask]))
                    else:
                        fraction_of_positives.append(np.nan)
                        mean_predicted_value.append(np.nan)

                fraction_of_positives = np.array(fraction_of_positives)
                mean_predicted_value = np.array(mean_predicted_value)

                plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Perfect Calibration')
                plt.plot(mean_predicted_value, fraction_of_positives, 's-', label='Tree-based Model (Calibrated)')
                plt.title(f'Calibration Curve for {subset_name}')
                plt.xlabel('Mean Predicted Probability')
                plt.ylabel('Fraction of True Positives')
                plt.legend()
                plt.grid(True, linestyle='--', alpha=0.6)
            else:
                logger.warning(f"Cannot plot calibration curve for {subset_name}: y_true and predictions have different lengths.")
                plt.text(0.5, 0.5, "Inconsistent calibration data", ha='center', va='center', fontsize=12, color='red')
                plt.title(f'Calibration Curve for {subset_name} (Inconsistent Data)')

            plt.tight_layout()
            plt.show()
        else:
            print("\n(Insufficient data for distribution or calibration visualizations.)")


        if cluster_coherence_alerts:
            cluster_ids = [a['cluster'] for a in cluster_coherence_alerts]
            gaps_in_percentage = [a['gap'] * 100 for a in cluster_coherence_alerts]

            plt.figure(figsize=(8, 5))
            sns.barplot(x=cluster_ids, y=gaps_in_percentage, palette='viridis')
            plt.title(f'Tree-based Model vs. Rule Coherence Gap per Cluster for {subset_name}')
            plt.xlabel('Cluster ID')
            plt.ylabel('Coherence Gap (%)')
            plt.axhline(y=SystemConfiguration.COHERENCE_ALERT_THRESHOLD * 100, color='r', linestyle='--', label='Alert Threshold')
            plt.legend()
            plt.grid(axis='y', linestyle='--', alpha=0.6)
            plt.tight_layout()
            plt.show()
        else:
            print("\n(No cluster coherence alerts to visualize.)")


        logger.info(f"✅ Fidelity and coherence analysis completed for {subset_name}")
        return {
            'coherence_alerts': cluster_coherence_alerts
        }

# --- Module 8: Fidelity & Coherence Analyzer ---

class FidelityCoherenceAnalyzer:
    """Calculates and presents global fidelity and coherence metrics of the model."""
    def __init__(self, config: SystemConfiguration):
        self.config = config

    def analyze_interpretability_measures(self, tree_model_pred_proba: np.ndarray,
                                            interp_feature_scores: np.ndarray,
                                            y_true: np.ndarray,
                                            pflz_cp_instance: Any,
                                            subset_name: str,
                                            cluster_interpretations_details: Dict):
        """
        Calculates and displays global fidelity and coherence metrics,
        as well as calibration diagnostics.
        """
        logger.info(f"Analyzing fidelity and coherence measures for {subset_name}...")

        normalized_interp_scores = interp_feature_scores[:, 0]

        global_mean_gap = np.mean(np.abs(tree_model_pred_proba - normalized_interp_scores))
        global_coherence_calc = 1 - global_mean_gap

        global_fidelity_calc = global_coherence_calc

        adaptive_threshold = np.percentile(normalized_interp_scores, self.config.PERCENTILE_COVERAGE)
        adaptive_coverage_calc = np.mean(normalized_interp_scores >= adaptive_threshold) * 100

        print(f"\n{'-'*70}")
        print(f"FIDELITY & COHERENCE SUMMARY FOR {subset_name}")
        print(f"{'-'*70}")
        print(f"📊 Global Coherence (1 - Mean Absolute Gap): {global_coherence_calc:.4f}")
        print(f"📊 Global Mean Absolute Gap: {global_mean_gap:.4f}")
        print(f"📊 Global Fidelity: {global_fidelity_calc:.4f}")
        print(f"📊 Adaptive Coverage ({self.config.PERCENTILE_COVERAGE}th percentile): {adaptive_coverage_calc:.2f}%")

        if global_mean_gap > self.config.COHERENCE_ALERT_THRESHOLD:
            print("🚨 GLOBAL COHERENCE ALERT: The gap between model predictions and rule interpretations is significant.")
            print("    A more in-depth analysis of cluster-level discrepancies is recommended.")
        else:
            print("✅ SATISFACTORY GLOBAL COHERENCE: Model predictions are well-aligned with rule interpretations.")

        X_df_for_plotting = pd.DataFrame()
        if hasattr(pflz_cp_instance, 'X_original_fit') and pflz_cp_instance.X_original_fit is not None and pflz_cp_instance.feature_names is not None:
            if pflz_cp_instance.X_original_fit.shape[1] == len(pflz_cp_instance.feature_names):
                X_df_for_plotting = pd.DataFrame(pflz_cp_instance.X_original_fit, columns=pflz_cp_instance.feature_names)
            else:
                logger.warning(f"Dimension mismatch between X_original_fit ({pflz_cp_instance.X_original_fit.shape[1]} cols) and feature_names ({len(pflz_cp_instance.feature_names)} names). Skipping DataFrame creation for cluster analysis.")
        else:
            logger.warning("Original X data or feature names are not available in PFLZ_CP instance. Skipping cluster visualizations.")

        if not X_df_for_plotting.empty and pflz_cp_instance.cluster_centers_.shape[0] > 0:
            RiskClusterAnalyzer.analyze_risk_clusters(
                X=X_df_for_plotting,
                centroids=pflz_cp_instance.cluster_centers_,
                predictions=tree_model_pred_proba,
                subset_name=subset_name,
                selected_features=pflz_cp_instance.feature_names,
                pflz_cp=pflz_cp_instance,
                cluster_interpretations=cluster_interpretations_details,
                y_true=y_true,
                normalized_interp_scores=normalized_interp_scores
            )
        else:
            print(f"\n(Cannot generate detailed cluster analysis for {subset_name} because original X data or centroids are not valid.)")

        logger.info(f"✅ Fidelity and coherence analysis completed for {subset_name}")

# --- Main Pipeline ---

class MainPipeline:
    """Main pipeline for clinical integration, orchestrating all steps."""

    def __init__(self):
        self.config = SystemConfiguration()
        self.data_manager = DataManager()
        self.X_train = None
        self.y_train = None
        self.all_feature_names = None
        self.selected_features = None

        self.config.validate_configuration()

        logger.info("🌟 Main pipeline initialized.")

    def _initialize_and_load_data(self):
        """Data initialization: mounts Drive, locates and loads datasets."""
        self.data_manager.mount_drive()
        self.data_manager.locate_data_folder()
        self.X_train, self.y_train, self.all_feature_names = self.data_manager.load_data()

        if self.X_train is None or self.y_train is None:
            raise ValueError("Data loading failed")

        self.selected_features = [col for col in self.X_train.columns if col != 'A']
        SystemConfiguration.NB_FEATURES = len(self.selected_features)
        SystemConfiguration.INDICES_FEATURES = list(range(SystemConfiguration.NB_FEATURES))

        logger.info(f"✅ Data loaded: {self.X_train.shape[0]} patients, {len(self.selected_features)} selected features (excluding 'A')")

        print("\n" + "="*50)
        print("SYSTEM CONFIGURATION")
        print("="*50)
        print("🏥 FIXED CLINICAL THRESHOLDS:")
        for level, threshold in self.config.CLINICAL_RISK_THRESHOLDS.items():
            print(f"    • {level.upper()} RISK: {threshold}%")
        print("\n🔧 Functions:")
        print("    • 🌟 Adaptive thresholds for coverage (60th percentile)")
        print("    • 🌟 Normalization of interpretability scores (0-1)")
        print("    • 🌟 Post-training calibration (isotonic)")
        print("    • 🌟 Alert system for inconsistencies (threshold: 15%)")
        print("    • 🌟 Weighting of interpretability features (increased to 15)")
        print("    • 🌟 Fixed clinical thresholds for comparability")
        print("    • 🌟 Detailed logging and diagnostics")
        print("    • 🌟 Class weighting to improve recall")
        print("    • 🌟 Classification threshold adjusted to 0.2 to improve recall")
        print("    • 🌟 Number of PFL rules: 7")
        print("    • 🌟 All features (except 'A') used for rules")
        print("    • 🌟 Rule interval widths adjusted (alpha=1.5)")
        print("\n📊 DEFINITIONS:")
        print("    • y: Observed outcome - 1 (72h Mortality), 0 (Survival)")
        print("    • A: Antibiotic treatment - 1 (Single), 0 (Multiple)")
        print("    • **Accuracy**: Proportion of correct predictions (true positives + true negatives) relative to the total number of observations.")
        print("    • **Balanced Accuracy**: The average recall (sensitivity) obtained on each class. Useful for imbalanced datasets.")
        print("    • **Kappa (Cohen's Kappa)**: A measure of agreement between predictions and ground truth, correcting for agreement that might be due to chance. A value of 1 indicates perfect agreement, 0 a random agreement.")
        print("    • **MCC (Matthews Correlation Coefficient)**: A measure of the quality of binary classifiers, even with imbalanced classes. Ranges from -1 (total disagreement) to +1 (perfect agreement), with 0 being a random result.")
        print("    • **Precision**: The proportion of true positives among all positive predictions. Answers the question: \"Of my positive predictions, how many are actually positive?\"")
        print("    • **Recall (Sensitivity)**: The proportion of true positives among all actual positive cases (those that should have been predicted positive). Answers the question: \"Of all actual positive cases, how many did I correctly identify?\"")
        print("    • **F1-Score**: The harmonic mean of precision and recall. A good balance between the two metrics, useful when seeking a balance between false positives and false negatives.")
        print("    • **Specificity**: The proportion of true negatives among all actual negative cases. Measures the model's ability to correctly identify negative cases.")
        print("    • **PR-AUC (Area Under the Precision-Recall Curve)**: Area under the precision-recall curve. Useful for evaluating models on highly imbalanced datasets, as it focuses on the performance of the minority class.")
        print("    • **ROC-AUC (Area Under the Receiver Operating Characteristic Curve)**: Area under the ROC curve. Indicates the model's ability to distinguish between classes. A value of 1 is perfect, 0.5 is random.")
        print("    • **Log Loss (Logarithmic Loss)**: Measures the performance of a classification model where the prediction is a probability. Heavily penalizes incorrect predictions made with high confidence.")
        print("    • **Brier Score**: Measures the accuracy of probabilistic predictions. A value of 0 indicates perfect predictions, 1 the worst possible score.")
        print("    • **FDR (False Discovery Rate)**: The proportion of false discoveries (false positives) among all cases predicted as positive. FDR = FP / (FP + TP).")
        print("    • **FOR (False Omission Rate)**: The proportion of false omissions (false negatives) among all cases predicted as negative. FOR = FN / (FN + TN).")
        print("    • **ECE (Expected Calibration Error)**: Measures the difference between the model's predicted confidence and its actual accuracy. A low ECE indicates that the model's predicted probabilities are reliable.")
        print("    • **Calibration Slope**: In a calibration plot, this is the slope of the regression line of observed proportions against predicted probabilities. A slope of 1 indicates perfect calibration.")
        print("    • **Coherence**: The degree of similarity between the tree-based model's predictions and the interpretations based on fuzzy rules. A high value indicates good interpretability of the model by its associated rules.")
        print("    • **Post-training Calibration (isotonic)**: A technique to adjust a model's output probabilities so that they better reflect the true probabilities of the event. Isotonic regression is non-parametric and flexible, seeking a monotonic function to map raw probabilities to calibrated probabilities.")
        print("    • **Adaptive Coverage**: Percentage of patients whose predictions can be considered well-interpreted by the rules, based on an adaptive interpretability score threshold (here, the 75th percentile). The higher this percentage, the more interpretable the model is for a large portion of cases.")

        print("="*50)
        print(f"\n=== SELECTED FEATURES ({self.config.NB_FEATURES}) ===")
        for i, name in enumerate(self.selected_features):
            print(f"    {i+1}. {name}")

    def _load_pflz_module(self):
        """Loads the PFLZ module from the specified directory."""
        global PFLZ_CP
        global _PFLZ_MODULE_LOADED
        try:
            loaded_pflz_class = PFLModuleLoader.load_pflz_cp_module(self.data_manager.mydrive_path)
            if loaded_pflz_class is None:
                raise ImportError("PFLZ class could not be found or loaded.")
            PFLZ_CP = loaded_pflz_class
            _PFLZ_MODULE_LOADED = True
            logger.info("✅ PFLZ_CP module loaded and validated")
        except Exception as e:
            logger.error(f"🚨 Error loading PFLZ_CP: {e}")
            _PFLZ_MODULE_LOADED = False
            raise


    def _prepare_subsets(self):
        """Prepares data subsets based on the 'A' variable."""
        logger.info("Preparing data subsets.")
        mask_A0 = self.X_train['A'] == 0
        mask_A1 = self.X_train['A'] == 1

        X_train_no_A = self.X_train.drop('A', axis=1)

        X_train_A0 = X_train_no_A[mask_A0]
        y_train_A0 = self.y_train[mask_A0]
        X_train_A1 = X_train_no_A[mask_A1]
        y_train_A1 = self.y_train[mask_A1]
        X_train_all = X_train_no_A
        y_train_all = self.y_train

        logger.info(f"Dataset A=0: {len(X_train_A0)} patients ({np.mean(y_train_A0):.1%} mortality)")
        logger.info(f"Dataset A=1: {len(X_train_A1)} patients ({np.mean(y_train_A1):.1%} mortality)")
        logger.info(f"Dataset ALL: {len(X_train_all)} patients ({np.mean(y_train_all):.1%} mortality)")
        return {
            'A0': (X_train_A0, y_train_A0),
            'A1': (X_train_A1, y_train_A1),
            'ALL': (X_train_all, y_train_all)
        }

    def _display_final_summary(self, results: Dict, cluster_analyses: Dict):
        """Displays a final summary of system results and functions."""
        print("\n" + "="*60)
        print("FINAL SUMMARY - PFL-TREE-BASED SYSTEM")
        print("="*60)

        print("\n🌟 Functions:")
        print("    • ✨ Adaptive thresholds for Coverage (60th percentile)")
        print("    • 🌟 Normalization of interpretability scores (0-1)")
        print("    • ✨ Fixed clinical thresholds for cross-dataset comparability")
        print("    • 🌟 Post-training calibration (ECE reduction)")
        print("    • ✨ Alert system for inconsistencies (> 10% per cluster)")
        print("    • 🌟 Increased weighting of interpretability features for the tree-based model (15)")
        print("    • ✨ Denormalized rules for increased clinical readability (alpha=1.5)")
        print("    • 🌟 Detailed diagnostics and comprehensive logging")
        print("    • ✨ Class weighting to improve recall")
        print("    • 🌟 Classification threshold adjusted to 0.2 to improve recall")
        print("    • ✨ Number of PFL rules = 7")
        print("    • 🌟 All features (except 'A') used for rules")

        print("\n✨ RESULTS PER DATASET:")
        for subset_name in ["A=All", "A=0", "A=1"]:
            subset_results = results[subset_name]['results']
            specific_analysis_results = cluster_analyses.get(subset_name, {}).get('coherence_alerts', [])

            print(f"\n 🌟 {subset_name}:")
            print(f"    • Tree-based model ROC-AUC: {np.nanmean(subset_results['metrics']['roc_auc'])*100:.2f}%")
            print(f"    • Raw ECE: {subset_results.get('ece_raw', np.nan):.4f}")
            print(f"    • Calibrated ECE: {subset_results.get('ece_calibrated', np.nan):.4f}")
            print(f"    • ECE Improvement: {subset_results.get('ece_improvement', np.nan):.4f}")
            print(f"    • Cluster inconsistency alerts: {len(specific_analysis_results)}")
            print(f"    • Recall: {np.nanmean(subset_results['metrics']['recall'])*100:.2f}%")

        print("\n✨ CLINICAL IMPACT:")
        print("    • Better calibrated probabilities → Increased confidence in tree-based model predictions.")
        print("    • Fixed thresholds → Comparability between patient populations.")
        print("    • Adaptive coverage → More patients with reliable interpretations (detailed in the dedicated analysis).")
        print("    • Cluster-level alert system → Targeted detection of complex cases for the tree-based model.")
        print("    • Detailed diagnostics → Debugging and continuous improvement.")
        print("    • Denormalized rules → Facilitated clinical decision-making.")
        print("    • Improved recall → Better detection of mortality cases.")

        print("\n🌟 USAGE RECOMMENDATIONS:")
        print("    1. ✨ Excellent Coherence (< 5% gap per cluster): Use rules as the main guide, with the tree-based model confirming.")
        print("    2. 🌟 Good Coherence (5-10% gap per cluster): Combine rules and tree-based model with caution, enhanced surveillance required.")
        print("    3. ✨ Poor Coherence (> 10% gap per cluster): Prioritize tree-based model predictions and investigate complex interactions.")
        print("    4. 🌟 Regularly monitor the tree-based model's calibration with new patients.")
        print("    5. ✨ Periodically recalibrate if a degradation in ECE is observed for the tree-based model.")

        print("\n" + "="*60)
        print("✨ SYSTEM READY FOR CLINICAL USE")
        print("="*60)
        logger.info("🌟 Final summary displayed successfully.")


    def execute(self):
        """Executes the entire pipeline: loading, training, evaluation, and analysis."""
        try:
            logger.info("Starting pipeline execution.")

            self._initialize_and_load_data()
            self._load_pflz_module()

            datasets = self._prepare_subsets()

            trainer = ModelTrainer(self.config)
            fidelity_coherence_analyzer = FidelityCoherenceAnalyzer(self.config)

            print("\n" + "="*50)
            print("TREE-BASED MODEL TRAINING AND EVALUATION")
            print("="*50)

            subsets_info = [
                ("A=0", "Multiple antibiotics", datasets['A0']),
                ("A=1", "Single antibiotic", datasets['A1']),
                ("A=All", "Combined dataset", datasets['ALL'])
            ]

            all_results = {}
            all_analysis_clusters = {}

            for subset_name, description, (X_data_full, y_data) in subsets_info:
                print(f"\n🔧 Processing for {subset_name} ({description})...")

                X_data_selected_features = X_data_full

                results_train_eval = trainer.train_and_evaluate(
                    X_data_selected_features, y_data, subset_name, self.selected_features
                )
                all_results[subset_name] = results_train_eval

                print(f"\n✨ LAUNCHING DETAILED FIDELITY & COHERENCE ANALYSIS FOR {subset_name}...")

                recalculated_interp_features, _, _ = InterpretabilityCalculator.calculate_interpretability_features(
                    results_train_eval['pflz_cp'],
                    X_data_selected_features.values,
                    y_data
                )

                fidelity_coherence_analyzer.analyze_interpretability_measures(
                    tree_model_pred_proba=results_train_eval['predictions'],
                    interp_feature_scores=recalculated_interp_features,
                    y_true=y_data,
                    pflz_cp_instance=results_train_eval['pflz_cp'],
                    subset_name=subset_name,
                    cluster_interpretations_details=results_train_eval['interpretations']
                )

                print(f"\n" + "="*50)
                print(f"RISK CLUSTER ANALYSIS FOR {subset_name}")
                print(f"=")

                all_analysis_clusters[subset_name] = RiskClusterAnalyzer.analyze_risk_clusters(
                    X_data_selected_features,
                    results_train_eval['centroids'],
                    results_train_eval['predictions'],
                    subset_name,
                    self.selected_features,
                    results_train_eval['pflz_cp'],
                    results_train_eval['interpretations'],
                    results_train_eval['raw_predictions'],
                    y_true=y_data,
                    normalized_interp_scores=recalculated_interp_features[:, 0]
                )

                if subset_name == "A=All":
                    logger.info("Calculating and storing the global rule synthesis for 'A=All'...")

                    pflz_cp_instance_all = results_train_eval['pflz_cp']
                    interpretations_clusters_all = results_train_eval['interpretations']

                    if not hasattr(pflz_cp_instance_all, 'cluster_centers_') or pflz_cp_instance_all.cluster_centers_ is None or pflz_cp_instance_all.cluster_centers_.shape[0] == 0:
                        logger.warning("Missing PFLZ centroids for 'A=All', cannot synthesize global rule.")
                    else:
                        cluster_centers_all = pflz_cp_instance_all.cluster_centers_
                        num_clusters_all = cluster_centers_all.shape[0]
                        num_features_all = cluster_centers_all.shape[1]

                        rule_widths_all = pflz_cp_instance_all.calc_rule_widths()
                        if rule_widths_all is None or rule_widths_all.shape[0] == 0:
                            logger.warning("Rule widths not available, using defaults.")
                            rule_widths_all = np.full((num_clusters_all, num_features_all), 0.5)

                        X_original_fit_all = pflz_cp_instance_all.X_original_fit
                        if X_original_fit_all is None or X_original_fit_all.shape[0] == 0:
                            logger.warning("Missing X_original_fit for 'A=All', uniform cluster weighting.")
                            cluster_weights_all = np.ones(num_clusters_all) / num_clusters_all
                        else:
                            membership_values_all = pflz_cp_instance_all.calc_mf(X_original_fit_all, normalized=True)
                            if membership_values_all is None or membership_values_all.shape[1] == 0:
                                cluster_weights_all = np.ones(num_clusters_all) / num_clusters_all
                            else:
                                raw_cluster_weights_all = np.sum(membership_values_all, axis=0)
                                cluster_weights_all = raw_cluster_weights_all / np.sum(raw_cluster_weights_all)

                        meta_centroid_all = np.zeros(num_features_all)
                        global_mortality_prob_rules_all = 0.0
                        meta_widths_all = np.zeros(num_features_all)

                        for i in range(num_clusters_all):
                            weight = cluster_weights_all[i] if i < len(cluster_weights_all) else (1.0 / num_clusters_all)
                            meta_centroid_all += cluster_centers_all[i] * weight
                            global_mortality_prob_rules_all += interpretations_clusters_all[i].get('conditional_prob', 0.0) * weight
                            if rule_widths_all.shape[0] > i:
                                meta_widths_all += rule_widths_all[i] * weight
                            else:
                                meta_widths_all += np.full(num_features_all, 0.5) * weight

                        if np.any(meta_widths_all <= 0):
                            meta_widths_all = np.full(num_features_all, 0.5)

                        global_formatted_rule_all = InterpretabilityCalculator._generate_centroid_rule(
                            centroid=meta_centroid_all,
                            feature_names=self.selected_features,
                            scalers=self.config.SCALERS,
                            alpha=1.5,
                            widths=meta_widths_all
                        )

                        all_results[subset_name]['global_rule_synthesis'] = {
                            'meta_centroid': meta_centroid_all,
                            'global_mortality_prob_rules': global_mortality_prob_rules_all,
                            'meta_widths': meta_widths_all,
                            'formatted_rule': global_formatted_rule_all,
                        }
                        logger.info("✅ Global rule synthesis calculated and stored in results.")

            self._display_final_summary(all_results, all_analysis_clusters)

            logger.info("🌟 Pipeline executed successfully.")
            return all_results, all_analysis_clusters

        except Exception as e:
            logger.error(f"🚨 Error in the pipeline: {str(e)}")
            traceback.print_exc()
            sys.exit(1)

# --- Main entry point ---

def display_final_global_rule(
    all_results_from_pipeline: Dict[str, Any],
    pipeline_config: Any
):
    """
    Displays a single global rule for the entire dataset,
    EXCLUSIVELY by reading the synthesized results already calculated and stored by the pipeline.
    This function does NOT redo any heavy calculations or loading.
    """
    logger.info("Starting streamlined display of the unique global rule.")
    print("\n" + "="*70)
    print("✨ UNIQUE GLOBAL RULE FOR THE ENTIRE DATASET ✨")
    print("="*70)

    results_all_data = all_results_from_pipeline.get('A=All')

    if not results_all_data or 'global_rule_synthesis' not in results_all_data:
        logger.error("Global rule synthesis is not available. Cannot display.")
        print("\n--- 🚨 ERROR: The global rule was not synthesized by the pipeline. Cannot display. ---")
        return

    global_synthesis_data = results_all_data['global_rule_synthesis']
    global_formatted_rule = global_synthesis_data['formatted_rule']
    global_mortality_prob_rules = global_synthesis_data['global_mortality_prob_rules']

    calibrated_predictions_all = results_all_data['predictions']
    raw_predictions_all = results_all_data['raw_predictions']
    avg_calibrated_pred_model = np.mean(calibrated_predictions_all) if len(calibrated_predictions_all) > 0 else 0.0
    avg_raw_pred_model = np.mean(raw_predictions_all) if len(raw_predictions_all) > 0 else 0.0

    normalized_interp_scores_all = results_all_data['diagnostics'].get('normalized_interpretability_scores_original')

    if normalized_interp_scores_all is None or len(normalized_interp_scores_all) == 0:
        logger.error("Normalized interpretability scores missing for global coherence. Cannot calculate gap.")
        global_coherence_gap = np.nan
        global_fidelity = np.nan
    else:
        min_len = min(len(calibrated_predictions_all), len(normalized_interp_scores_all))
        if min_len == 0:
            global_coherence_gap = np.nan
            global_fidelity = np.nan
        else:
            global_coherence_gap = np.mean(np.abs(calibrated_predictions_all[:min_len] - normalized_interp_scores_all[:min_len]))
            global_fidelity = 1 - global_coherence_gap


    total_patient_count = len(calibrated_predictions_all)

    print(f"• **Calibrated risk (Tree-based model)**: {avg_calibrated_pred_model:.2%}")
    print(f"• **Raw risk (Tree-based model)**: {avg_raw_pred_model:.2%}")
    print(f"• **P(mortality|Global Fuzzy Rule)**: {global_mortality_prob_rules:.2%}")
    print(f"• **Fidelity (Tree-based model vs. Global Rules)**: {global_fidelity:.4f}")

    global_rule_risk_level = "N/A"
    if (global_mortality_prob_rules * 100) < pipeline_config.CLINICAL_RISK_THRESHOLDS['LOW']:
        global_rule_risk_level = "LOW"
    elif (global_mortality_prob_rules * 100) <= pipeline_config.CLINICAL_RISK_THRESHOLDS['MEDIUM']:
        global_rule_risk_level = "MEDIUM"
    else:
        global_rule_risk_level = "HIGH"
    print(f"• **Global clinical recommendation**: {global_rule_risk_level}")

    print(f"• **Total patient count in global dataset**: {total_patient_count:,}")
    print(f"• **Percentage of global subset**: 100.0%")
    print(f"• **Global clinical rule**: **{global_formatted_rule}**")

    coherence_status_display = ""
    if not np.isnan(global_coherence_gap):
        if global_coherence_gap < (pipeline_config.COHERENCE_ALERT_THRESHOLD / 2):
            coherence_status_display = "✅ Excellent"
        elif global_coherence_gap < pipeline_config.COHERENCE_ALERT_THRESHOLD:
            coherence_status_display = "✨ Good"
        else:
            coherence_status_display = "🚨 Poor - In-depth analysis required"
        print(f"• **Coherence (Rule vs. Tree-based model, Global)**: {coherence_status_display} (gap: {global_coherence_gap:.1%})")

        if global_coherence_gap > pipeline_config.COHERENCE_ALERT_THRESHOLD:
            print("\n    🚨 GLOBAL ALERT: A significant gap between model predictions and the global rule has been detected.")
            print("      This indicates that the 'average profile' of the rule does not fully capture the complexity of the prediction model.")
            print("      A review of the specific cluster alerts (displayed previously) is recommended to target problematic subgroups.")
    else:
        print(f"• **Coherence (Rule vs. Tree-based model, Global)**: Not calculable (missing or inconsistent data)")


    print("\n" + "="*70)
    print("✨ GLOBAL RULE SYNTHESIZED AND DISPLAYED SUCCESSFULLY")
    print("="*70)

    logger.info("Unique global rule synthesized and displayed successfully.")


if __name__ == "__main__":
    try:
        warnings.filterwarnings('ignore')
        logger.info("🌟 Starting PFL system.")

        pipeline = MainPipeline()
        results, analyses = pipeline.execute()

        logger.info("✨ Pipeline execution completed successfully!")

        display_final_global_rule(
            all_results_from_pipeline=results,
            pipeline_config=pipeline.config
        )

        logger.info("🌟 Full process completed.")

    except Exception as e:
        logger.error(f"🚨 Fatal error: {e}")
        traceback.print_exc()
        sys.exit(1)
```
